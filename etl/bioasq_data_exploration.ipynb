```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_excel("../data/BioASQ_extracted.xlsx")
df.head(10)
```

```python
df['type'].value_counts()
```

```python
df.groupby("type").agg({"num_documents": "sum"})
```

```python
px.histogram(data_frame=df, x="type", y="num_documents", color="type", title="Number of documents per type")
```

```python
px.box(data_frame=df, x="type", y="num_documents", color="type", title="Distribution of number of documents per type")
```

```python
df_sampled = df.groupby("type", group_keys=False).apply(lambda x: x.sample(n=2, random_state=42))
```

```python
df_sampled['pubmed_ids'] = df_sampled['pubmed_ids'].apply(lambda x: list(str(x).strip() for x in x.split(",")))
```

```python
df_sampled
```

```python
df_pub_ids = df_sampled.explode(["pubmed_ids"])
df_pub_ids
```

```python
df_pub_ids['pubmed_ids'].drop_duplicates().to_csv("../data/pmdids_to_fetct.txt", index=False, header=False)
```
